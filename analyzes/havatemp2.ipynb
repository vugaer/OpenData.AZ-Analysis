## Load All Datasets

### Subtask:
Dynamically load all 19 CSV files (data0.csv to data18.csv) into pandas DataFrames, guessing the separator (',' or ';') for each file.


**Reasoning**:
I need to define a function to robustly load CSV files, guessing the separator, and then iterate through all dataset indices to load each file, display its basic information, and make it globally accessible as instructed.



````python
def load_dataframe_robustly(index):
  filepath = f'./datasets/data{index}.csv'
  try:
    # Try comma separator first
    df = pd.read_csv(filepath, sep=',')
    # Check if it loaded as a single column without ',' in its name (potential wrong separator)
    if df.shape[1] == 1 and df.columns[0].count(',') > 0:
      df = pd.read_csv(filepath, sep=';')
  except Exception as e:
    print(f"Error loading data{index}.csv: {e}")
    try:
      # If comma failed, try semicolon
      df = pd.read_csv(filepath, sep=';')
      # If it's still a single column and semicolon was the only option, it might be correctly parsed.
      # But if the column name contains ';', it indicates it was not properly separated.
      if df.shape[1] == 1 and df.columns[0].count(';') > 0:
        print(f"Warning: data{index}.csv still loaded as single column with ';' in name. Might need manual inspection.")
    except Exception as e_semicolon:
      print(f"Error loading data{index}.csv with semicolon: {e_semicolon}")
      return None

  # Final check: if it's still a single column and the first column contains either ',' or ';'
  # it's likely that the separator guessing failed. This is a heuristic.
  if df.shape[1] == 1 and (df.columns[0].count(',') > 0 or df.columns[0].count(';') > 0):
    print(f"Warning: data{index}.csv might not be parsed correctly. It has only one column: {df.columns[0]}")
    return df # Return it anyway for inspection, but with a warning

  return df

dataframes = {}
for i in range(19):
  df_name = f'df{i}'
  globals()[df_name] = load_dataframe_robustly(i)
  dataframes[df_name] = globals()[df_name]

  print(f"\n{'='*20} Dataframe {df_name} Overview {'='*20}")
  if dataframes[df_name] is not None:
    print(f"First 5 rows of {df_name}:")
    display(dataframes[df_name].head())
    print(f"\nInfo for {df_name}:")
    dataframes[df_name].info()
    print(f"\nDescriptive statistics for {df_name}:")
    display(dataframes[df_name].describe(include='all'))
  else:
    print(f"Could not load dataframe {df_name}.")
  print("-" * 50)
````

## Cross-Dataset Exploration and Analysis

### Subtask:
Combine relevant datasets to analyze the relationship between revenue, passenger turnover, and expenses across different transport modes over time.


**Reasoning**:
To analyze the relationship between revenue, passenger turnover, and expenses across different transport modes, I will identify common transport mode columns, transform the relevant dataframes (`df0`, `df3`, `df5`) into a long format, merge them, and then calculate the requested ratios for further analysis.



````python
import numpy as np

# 1. Define common transport mode columns
common_transport_modes = list(set(df0.columns) & set(df3.columns) & set(df5.columns))
# Exclude 'year' and 'cəmi' as specified
common_transport_modes = [col for col in common_transport_modes if col not in ['year', 'cəmi']]

print(f"Common transport mode columns: {common_transport_modes}")

# 2. Transform each DataFrame to a long format using pd.melt()
# df0 (revenue from passenger transport)
df0_melted = pd.melt(df0, id_vars=['year'], value_vars=common_transport_modes,
                     var_name='transport_mode', value_name='revenue')
print("df0_melted created.")

# df3 (passenger turnover)
df3_melted = pd.melt(df3, id_vars=['year'], value_vars=common_transport_modes,
                     var_name='transport_mode', value_name='passenger_turnover')
print("df3_melted created.")

# df5 (transport expenses)
df5_melted = pd.melt(df5, id_vars=['year'], value_vars=common_transport_modes,
                     var_name='transport_mode', value_name='expenses')
print("df5_melted created.")

# 3. Perform a series of merges
df_combined = pd.merge(df0_melted, df3_melted, on=['year', 'transport_mode'], how='outer')
df_combined = pd.merge(df_combined, df5_melted, on=['year', 'transport_mode'], how='outer')
print("Dataframes merged into df_combined.")

# 4. Display the first 5 rows of df_combined
print("\nFirst 5 rows of df_combined:")
display(df_combined.head())

# 5. Print the information about df_combined
print("\nInfo for df_combined:")
df_combined.info()

# 6. Calculate revenue_per_passenger_turnover
df_combined['revenue_per_passenger_turnover'] = df_combined['revenue'] / df_combined['passenger_turnover']
# Handle division by zero (inf values)
df_combined['revenue_per_passenger_turnover'].replace([np.inf, -np.inf], np.nan, inplace=True)
print("\nCalculated 'revenue_per_passenger_turnover'.")

# 7. Calculate expenses_per_passenger_turnover
df_combined['expenses_per_passenger_turnover'] = df_combined['expenses'] / df_combined['passenger_turnover']
# Handle division by zero (inf values)
df_combined['expenses_per_passenger_turnover'].replace([np.inf, -np.inf], np.nan, inplace=True)
print("Calculated 'expenses_per_passenger_turnover'.")

# 8. Display the descriptive statistics of the newly calculated columns, grouped by 'transport_mode'
print("\nDescriptive statistics of new ratios grouped by transport_mode:")
display(df_combined.groupby('transport_mode')[['revenue_per_passenger_turnover', 'expenses_per_passenger_turnover']].describe())
````
